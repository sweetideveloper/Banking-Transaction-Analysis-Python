# Data Cleaning

```python
df.isnull().sum()
```

```python
df.drop_duplicates()
```

```python
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])
```

```python
df['Month'] = df['TransactionDate'].dt.month
```

# Visualization

## Histogram for transaction amount

```python
plt.figure()
df['TransactionAmount'].hist(bins=30)
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
type_analysis = df.groupby('TransactionType')['TransactionAmount'].agg(['sum','mean','count'])
type_analysis
```

```python
type_analysis['sum'].plot(kind='bar')
plt.title("Debit vs Credit Comparision")
plt.ylable("Total Amount")
plt.show()
```

```python
monthly = df.groupby('Month')['TransactionAmount'].sum()

monthly.plot(marker='<')
plt.title("Monthly Transaction Trend")
plt.xlable("Month")
plt.ylable("Amount")
plt.show()
```
